```python
import pandas as pd
import matplotlib.pyplot as plt

# CSV should contain: title, rating, revenue
file_name = "movies.csv"
df = pd.read_csv(file_name)

# Standardise column names and validate required fields
df.columns = [col.strip().lower() for col in df.columns]
required = {"title", "rating", "revenue"}
missing = required - set(df.columns)
if missing:
    raise ValueError(f"Missing required columns: {missing}")

# Clean and convert fields
df["title"] = df["title"].astype(str).str.strip()
df["rating"] = pd.to_numeric(df["rating"], errors="coerce")
df["revenue"] = pd.to_numeric(df["revenue"], errors="coerce")
df = df[df["title"].ne("") & df["title"].ne("nan")]

# Summaries
print("Number of movies:", len(df))
print("Average rating:", round(df["rating"].mean(), 2))
print("Total reported revenue:", round(df["revenue"].sum(), 2))
print("Average reported revenue:", round(df["revenue"].mean(), 2))

# Top 10 movies by rating and revenue
top_rated = df.dropna(subset=["rating"]).sort_values(
    "rating", ascending=False
).head(10)
top_revenue = df.dropna(subset=["revenue"]).sort_values(
    "revenue", ascending=False
).head(10)

print("\nTop-rated movies:")
print(top_rated[["title", "rating"]].to_string(index=False))
print("\nTop-revenue movies:")
print(top_revenue[["title", "revenue"]].to_string(index=False))

# Charts
if not top_rated.empty:
    top_rated.plot.bar(x="title", y="rating", legend=False,
                       title="Top Movies by Rating")
    plt.ylabel("Rating")
    plt.tight_layout()
    plt.show()

if not top_revenue.empty:
    top_revenue.plot.bar(x="title", y="revenue", legend=False,
                         title="Top Movies by Revenue")
    plt.ylabel("Revenue (dataset units)")
    plt.tight_layout()
    plt.show()
```
